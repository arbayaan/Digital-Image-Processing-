```python
import cv2
import matplotlib.pyplot as plt

# Load the available image in grayscale to define 'gray'
gray = cv2.imread('/content/images (1).png', cv2.IMREAD_GRAYSCALE)

# PDF / Normalized Histogram
hist = cv2.calcHist([gray], [0], None, [256], [0,256])

total_pixels = gray.size

pdf = hist / total_pixels

plt.figure(figsize=(10,5))

plt.plot(pdf)
plt.title("Normalized Histogram (PDF)")
plt.xlabel("Pixel Intensity")
plt.ylabel("Probability")

plt.xlim([0,256])
plt.grid()

# Save Output BEFORE calling show()
plt.savefig("3_2_Normalized_Histogram.png")
plt.show()
```
