```python
import cv2
import matplotlib.pyplot as plt

# Load the image and convert from BGR to RGB
img_bgr = cv2.imread('/content/images.jfif')
img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Original image
h, w = img.shape[:2]

# Resize image to 2 times its original size

# 1. Nearest Neighbor
nearest = cv2.resize(
    img, (w*2, h*2),
    interpolation=cv2.INTER_NEAREST
)

# 2. Bilinear
bilinear = cv2.resize(
    img, (w*2, h*2),
    interpolation=cv2.INTER_LINEAR
)

# 3. Bicubic
bicubic = cv2.resize(
    img, (w*2, h*2),
    interpolation=cv2.INTER_CUBIC
)

# Display Results
titles = ['Original Image',
          'Nearest Neighbor',
          'Bilinear Interpolation',
          'Bicubic Interpolation']

images = [img, nearest, bilinear, bicubic]

plt.figure(figsize=(14, 10))

for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()

# Save output images
cv2.imwrite('nearest_neighbor.png',
            cv2.cvtColor(nearest, cv2.COLOR_RGB2BGR))

cv2.imwrite('bilinear.png',
            cv2.cvtColor(bilinear, cv2.COLOR_RGB2BGR))

cv2.imwrite('bicubic.png',
            cv2.cvtColor(bicubic, cv2.COLOR_RGB2BGR))
```
