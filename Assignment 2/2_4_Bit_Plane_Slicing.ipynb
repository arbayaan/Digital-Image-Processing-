```python
import cv2
import matplotlib.pyplot as plt

# Bit Plane Slicing

# Load the image from the workspace
img = cv2.imread('/content/images.png')
# Convert BGR to RGB since cv2 loads images in BGR format
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# Create 8 bit planes
bit_planes = []

for i in range(8):
    plane = (gray >> i) & 1
    plane = plane * 255
    bit_planes.append(plane)

# Display Original + 8 Bit Planes
plt.figure(figsize=(15, 12))

plt.subplot(3, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

for i in range(8):
    plt.subplot(3, 3, i+2)
    plt.imshow(bit_planes[i], cmap='gray')
    plt.title(f'Bit Plane {i}')
    plt.axis('off')

plt.tight_layout()
plt.show()

# Save bit planes
for i in range(8):
    cv2.imwrite(f'bit_plane_{i}.png', bit_planes[i])
```
