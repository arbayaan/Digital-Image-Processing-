```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
# Reading image in RGB format for proper display in matplotlib
img = cv2.imread('/content/images.png')
if img is not None:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
else:
    # Create a placeholder if the image file cannot be read
    img = np.zeros((300, 300, 3), dtype=np.uint8)

# Gamma Correction
def gamma_correction(image, gamma):
    normalized = image / 255.0
    corrected = np.power(normalized, gamma)
    return np.uint8(corrected * 255)

# Different gamma values
gamma_05 = gamma_correction(img, 0.5)
gamma_10 = gamma_correction(img, 1.0)
gamma_15 = gamma_correction(img, 1.5)
gamma_20 = gamma_correction(img, 2.0)

# Display results
titles = ['Original Image', 'Gamma 0.5',
          'Gamma 1.0', 'Gamma 1.5',
          'Gamma 2.0']

images = [img, gamma_05, gamma_10,
          gamma_15, gamma_20]

plt.figure(figsize=(15, 10))

for i in range(5):
    plt.subplot(2, 3, i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()

# Save output
for title, image in zip(titles[1:], images[1:]):
    cv2.imwrite(title.replace(' ', '_') + '.png',
                cv2.cvtColor(image, cv2.COLOR_RGB2BGR))
```
